# Visualization of Listings Features

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
listings = pd.read_csv('../Preprocessing/listings_cleaned.csv')
```

```python
def feature_hist(feature):
    counts = (listings[feature].value_counts())

    plt.figure(figsize=(8, 5))
    plt.bar(counts.index, counts.value)

    plt.ylabel('count')
    plt.xlabel(feature)
    plt.xticklabels(counts.index, rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```
